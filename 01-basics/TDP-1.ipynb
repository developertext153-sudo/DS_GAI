```python
# Check the data type of 3.0
print(type(10)) # int
print(type(3.0)) # float    
print(type(10+20j)) # complex
print(type('Hello')) # str
print(type([1, 2, 3])) # list
print(type({'name':'Asabeneh'})) # dict
print(type({9.8, 3.14, 2.7})) # set 
print(type(3==3)) # bool
print(type(3>=3)) # bool
```
