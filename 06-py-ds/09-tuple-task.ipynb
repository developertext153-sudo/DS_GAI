```python
# Tuple Slicing
```

```python
mytuple=('one','two','three','four','five','six','seven','eight')
mytuple[0:3] # starting from 0 index to (3-1) i.e 2nd index
```

```python
mytuple[2:5] # starting from 2nd index to (5-1) i.e 4th index
```

```python
mytuple[:3] # starting from 0nd index to (3-1) i.e 2nd index
```

```python
mytuple[:2] # starting from 0nd index to (2-1) i.e 1st index
```

```python
mytuple[-3:] # starting from backward -3 index to last index
```

```python
mytuple[-2:] # starting from backward -2 index to last index
```

```python
mytuple[-1]
```

```python
mytuple[:]
```
