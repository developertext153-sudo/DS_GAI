```python
# Copy the element from s5 to s4
s4=s5.copy()
```

```python
s4.add(2)
print(s4)
```

```python
# Duplicates does not allowed in set
s4.add(2)
print(s4)
```

```python
# Remove all the element from set
s5.clear()
print(s5)
```

```python
# Deleting the set
del s5
# print(s5)
```

```python
# Removing the element from set
print(s4)
s4.remove(1+2j)
print(s4)
```

```python
# Printing s3 element
print(s3) 
```

```python
# Discard vs Remove
s3.discard('m')
```

```python
# s3.remove('m')
# deleting the random element from the set
s3.pop()
print(s3)
s3.pop()
print(s3)
```

```python
# set does not allow indexing and slicing
# s2.pop(3) # takes no arguments (1 given)
print(s2)
s2.pop()
print(s2)
```

```python
# Looping through set
for i in s2:
    print(i)

# Looping with enumerate in set
for i in enumerate(s2):
    print(i)
```
