```python
# Single line comments

letter='p'
print(letter)
print(len(letter))
greeting='Hello, World!'
print(greeting)
print(len(greeting))
sentence='I am enjoying 30 days of python'  
print(sentence)

```

```python
# MultiLine Strings
multiline_string='''I am enjoying 30 days of python
I just wonder what is next.'''
print(multiline_string)

# Another way of creating a multiline string is by using double quotes
multiline_string2="""I am enjoying 30 days of python    
I just wonder what is next."""
print(multiline_string2)
```

```python
# String Concatenation
first_Name='Asabeneh'
last_Name='Yetayeh' 
space=' '
full_Name=first_Name+space+last_Name
print(full_Name) 
```

```python
# Checking length of a string
print(len(full_Name))
print(len(first_Name))
print(len(last_Name))
print(len(full_Name))
```

```python
# Unpacking characters of a string
language='Python'
a,b,c,d,e,f=language # unpacking the characters of a string
print(a)
print(b)    
print(c)
print(d)
print(e)
print(f)
```

```python
# Accessing characters in strings by index
first_name='Prashant'
first_character=first_name[0]
print(first_character) # first character
second_character=first_name[1]
print(second_character) # second character
last_character=first_name[-1] # last character
print(last_character)   
```

```python
# If we want to access the from right end we can use negative indexing. The last character has an index of -1, the second last character has an index of -2 and so on.
first_name='Prashant'
last_character=first_name[-1] # last character
print(last_character)
second_last_character=first_name[-2] # second last character
print(second_last_character)
```

```python
# Slicing strings
first_name='Prashant'
print(first_name[:])
```

```python
first_three=first_name[0:3]
print(first_three) # first three characters
```

```python
last_three=first_name[-3:]
print(last_three) # last three characters
```

```python
# Skipping characters while slicing
first_name='Prashant'
print(first_name[0:7:2]) # prn
```

```python
# Escaping characters in strings
print('I hope everyone is enjoying the Python Challenge.\nAre you ?') # using \n to break line  
print('Days\tTopics\tExercises') # using \t to add tab space
print('Day 1\t3\t5')
print('Day 2\t3\t5')
print('Day 3\t3\t5')
print('Day 4\t3\t5')
print('Day 5\t3\t5')
print('This is backslash symbol (\\)') # To write a backslash
print('In every programming language it starts with \"Hello, World!\"') # To write double quote
```

```python
# String Methods
# capitalize(): Converts the first character of the string to uppercase
string='python for everyone'
print(string.capitalize()) # Python for everyone
```

```python
# count(): Returns the number of times a specified value occurs in a string
string='I am enjoying 30 days of python'
print(string.count('o')) # 3
```

```python
# endswith(): Returns true if the string ends with the specified value
string='I am enjoying 30 days of python'
print(string.endswith('python')) # True
```

```python
# expandtabs(): Sets the tab size of the string
string='H\te\tl\tl\to'
print(string.expandtabs()) # Hello
```

```python
# find(): Searches the string for a specified value and returns the position of where it was found
string='I am enjoying 30 days of python'
print(string.find('y')) # 17
```

```python
# format(): Formats specified values in a string    
first_name='Prashant'
last_name='Pandey'
country='India'
designation='Data Scientist'
print('I am {} {}, I am a {}, I live in {}'.format(first_name, last_name, designation, country)) # I am Prashant Pandey, I am a Data Scientist, I live in India 
```

```python
radius=10
pi=3.14
area='The are of circle with radius {} is {}'.format(radius, pi*radius**2)  
print(area) # 314.0
```

```python
# index(): Searches the string for a specified value and returns the position of where it was found
string='I am enjoying 30 days of python'
print(string.index('y')) # 17
```

```python
# isalnum(): Returns True if all characters in the string are alphanumeric
string='Python30'
print(string.isalnum()) # True

# isalpha(): Returns True if all characters in the string are in the alphabet
string='Python'
print(string.isalpha()) # True

string='30DaysOfPython'
print(string.isalpha()) # False, because there are numbers in the string

string='30 DaysOfPython'
print(string.isalnum()) # False, because there is a space in the string
```
